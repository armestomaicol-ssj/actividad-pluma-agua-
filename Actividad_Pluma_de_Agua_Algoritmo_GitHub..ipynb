## **Historia, ciencia y programación: la “pluma de agua”**

> **Relato de partida**  
> **¿Por qué los costeños le decimos “pluma” a la llave del agua?**  
>  
> En el Caribe colombiano, abrir la “pluma” no es solo una acción cotidiana: es una expresión cargada de historia, técnica y cultura. Aunque algunos se sorprendan o incluso se burlen del término, lo cierto es que decir “pluma” en vez de “llave” no es un error, sino una herencia viva que conecta nuestro lenguaje con siglos de evolución hidráulica.  
>  
> Mucho antes de que existieran los grifos modernos, el agua se distribuía mediante sistemas regulados con precisión. En España, por ejemplo, se usaba la **“pluma de agua”** como unidad de medida para calcular el caudal que se entregaba a cada hogar. Aproximadamente, una pluma equivalía a **0.025 litros por segundo**. En algunos lugares, ese flujo se regulaba mediante orificios calibrados y dispositivos llamados **plumeros**, que permitían controlar la salida del agua.  
>  
> Durante la época colonial, muchas de estas prácticas técnicas cruzaron el Atlántico y llegaron a ciudades como Cartagena, Barranquilla y Santa Marta. Con el tiempo, el término “pluma” dejó de ser solamente técnico y pasó al habla cotidiana como sinónimo de la llave del agua.  
>  
> Así, una palabra del uso diario conserva una memoria histórica, tecnológica y cultural.

## 2. Situación problema

Un hogar recibe una cierta cantidad de **plumas de agua**.  
Se sabe que:

- **1 pluma = 0.025 litros por segundo**

Con esta información, el estudiante debe construir un programa que permita calcular:

1. el caudal total en litros por segundo;
2. el consumo diario en litros;
3. el consumo mensual en litros (suponiendo 30 días);
4. una clasificación del consumo:
   - **Bajo**: menos de 1000 litros por día
   - **Medio**: entre 1000 y 3000 litros por día
   - **Alto**: más de 3000 litros por día

---

## 3. Análisis del problema

### 3.1 Entrada del sistema
El programa debe recibir:

- número de plumas asignadas al hogar.

### 3.2 Proceso
El programa debe:

- convertir el número de plumas en caudal total;
- transformar el caudal a consumo diario;
- transformar el consumo diario a consumo mensual;
- clasificar el nivel de consumo.

### 3.3 Salida
El programa debe mostrar:

- caudal total;
- consumo diario;
- consumo mensual;
- clasificación del consumo.

## 5. Algoritmo propuesto

### Algoritmo en lenguaje natural

## 6. Espacio para que el estudiante construya la solución

### Instrucciones
Complete el siguiente bloque de código a partir del algoritmo anterior.

- No copie una solución terminada sin comprenderla.

- Verifique que los resultados tengan sentido físico.

````python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# Grupo : N°7
# Curso: Fundamentos de Programación - Universidad de Sucre
# ==========================================================

# 1. Solicitar al usuario el número de plumas
plumas = int(input("Ingrese el número de plumas de agua asignadas al hogar: "))

# 2. Definir constantes del problema
caudal_pluma = 0.025      # litros por segundo que equivale una pluma
segundos_dia = 86400      # segundos que tiene un día (60 x 60 x 24)
dias_mes = 30             # días que se asumen en un mes

# 3. Calcular el caudal total
caudal_total = plumas * caudal_pluma

# 4. Calcular el consumo diario
consumo_diario = caudal_total * segundos_dia

# 5. Calcular el consumo mensual
consumo_mensual = consumo_diario * dias_mes

# 6. Clasificar el consumo
if consumo_diario < 1000:
    nivel_consumo = "Bajo"
elif consumo_diario <= 3000:
    nivel_consumo = "Medio"
else:
    nivel_consumo = "Alto"

# 7. Mostrar resultados
print("\n===== RESULTADOS =====")
print(f"Número de plumas asignadas:  {plumas}")
print(f"Caudal total:                {caudal_total} litros/segundo")
print(f"Consumo diario:              {consumo_diario} litros/día")
print(f"Consumo mensual:             {consumo_mensual} litros/mes")
print(f"Nivel de consumo:            {nivel_consumo}")
````

## 8. Extensión opcional

Si desea profundizar, puede ampliar el programa para que también permita:

- calcular el consumo semanal;
- comparar el consumo de varios hogares;
- estimar el costo del agua si se conoce una tarifa por litro o por metro cúbico;
- convertir litros a metros cúbicos;
- presentar los resultados en forma tabular.

> Estas extensiones no reemplazan la solución base; son una ampliación.

````python
# ==========================================================
# SECCIÓN 8: EXTENSIÓN DEL PROGRAMA
# ==========================================================

# --- DATOS BASE ---
plumas = int(input("Ingrese el número de plumas de agua asignadas al hogar: "))
caudal_pluma = 0.025
segundos_dia = 86400
dias_mes = 30
dias_semana = 7

# --- CÁLCULOS ---
caudal_total = plumas * caudal_pluma
consumo_diario = caudal_total * segundos_dia
consumo_semanal = consumo_diario * dias_semana
consumo_mensual = consumo_diario * dias_mes
consumo_m3 = consumo_mensual / 1000  # conversión a metros cúbicos

# --- CLASIFICACIÓN ---
if consumo_diario < 1000:
    nivel_consumo = "Bajo"
elif consumo_diario <= 3000:
    nivel_consumo = "Medio"
else:
    nivel_consumo = "Alto"

# --- COSTO ---
tarifa = float(input("Ingrese la tarifa por litro de agua (en pesos): "))
costo_mensual = consumo_mensual * tarifa

# --- COMPARACIÓN CON OTRO HOGAR ---
plumas2 = int(input("Ingrese el número de plumas del segundo hogar para comparar: "))
consumo_diario2 = (plumas2 * caudal_pluma) * segundos_dia
consumo_mensual2 = consumo_diario2 * dias_mes
consumo_m3_2 = consumo_mensual2 / 1000

if consumo_diario2 < 1000:
    nivel2 = "Bajo"
elif consumo_diario2 <= 3000:
    nivel2 = "Medio"
else:
    nivel2 = "Alto"

costo_mensual2 = consumo_mensual2 * tarifa

# --- TABLA DE RESULTADOS ---
print("\n" + "="*55)
print(f"{'CONCEPTO':<30} {'HOGAR 1':>10} {'HOGAR 2':>10}")
print("="*55)
print(f"{'Plumas asignadas':<30} {plumas:>10} {plumas2:>10}")
print(f"{'Caudal total (L/s)':<30} {caudal_total:>10.3f} {plumas2*caudal_pluma:>10.3f}")
print(f"{'Consumo diario (L)':<30} {consumo_diario:>10.1f} {consumo_diario2:>10.1f}")
print(f"{'Consumo semanal (L)':<30} {consumo_semanal:>10.1f} {consumo_diario2*7:>10.1f}")
print(f"{'Consumo mensual (L)':<30} {consumo_mensual:>10.1f} {consumo_mensual2:>10.1f}")
print(f"{'Consumo mensual (m³)':<30} {consumo_m3:>10.2f} {consumo_m3_2:>10.2f}")
print(f"{'Costo mensual ($)':<30} {costo_mensual:>10,.0f} {costo_mensual2:>10,.0f}")
print(f"{'Nivel de consumo':<30} {nivel_consumo:>10} {nivel2:>10}")
print("="*55)
````
